```python
#6.2
guess_me = 7
number = 1

while True:
    if number < guess_me:
        print("too low")
    elif number == guess_me:
        print("Found it")
        break
    else:
        print("oops")
        break
    number += 1
```

```python
#6.3
Guess_Me = 5
for number in range(10):
    if number < Guess_Me:
        print("too low")
    elif number == Guess_Me:
        print("Found it")
        break
    else:
        print("oops")
        break
```
